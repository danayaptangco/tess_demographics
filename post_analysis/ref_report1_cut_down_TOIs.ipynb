```python
import pandas as pd
import numpy as np
```

```python
df_no_cuts = pd.read_csv('./TOI_Mar2025_1pt5to4_R_with_extra_columns.csv')
df_no_cuts = df_no_cuts[df_no_cuts['pl_rade'] >=1.8]
df_no_cuts = df_no_cuts[df_no_cuts['pl_rade'] <=4]

print(len(df_no_cuts))
```

```python
df_period_cut = df_no_cuts[df_no_cuts['pl_orbper'] <= 25]
df_period_cut = df_period_cut[df_period_cut['pl_orbper'] >= 0.6]
print(len(df_period_cut))
```

```python
df_detectability_cut = pd.read_csv('./TOI_Mar2025_1pt5to4_R_detected_only.csv')

df_detectability_cut = df_detectability_cut[df_detectability_cut['pl_rade'] >=1.8]
df_detectability_cut = df_detectability_cut[df_detectability_cut['pl_rade'] <=4]
df_detectability_cut = df_detectability_cut[df_detectability_cut['pl_orbper'] <= 25]
df_detectability_cut = df_detectability_cut[df_detectability_cut['pl_orbper'] >= 0.6]
print(len(df_detectability_cut))
```
